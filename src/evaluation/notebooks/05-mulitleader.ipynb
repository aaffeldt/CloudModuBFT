```python
folder = "../../aws/benchmark/out/geobft/final-multileader-geobft-with-non-optimistic-global-sharing-2-buckets/" 
benchmarks = glob(f"{folder}*/")
```

```python

throughputs = process_throughput_benchmarks(benchmarks)
throughputs["files"] = throughputs["benchmark"]
throughputs["benchmark"] = "multileader non-optimistic-2-buckets"
throughputs["buckets"] = 2 
throughputs["num_peers"] = throughputs["num_peers"]/throughputs["number_of_clusters"]
processed_cpu_usage = process_cpu_usage(benchmarks)
processed_cpu_usage["num_peers"] = processed_cpu_usage["num_peers"]/processed_cpu_usage["number_of_clusters"]
processed_cpu_usage["files"] = processed_cpu_usage["benchmark"]
processed_cpu_usage["benchmark"] = "multileader non-optimistic-2-buckets"
processed_cpu_usage["buckets"] = 2 

bytes_sent = process_bytes_sent(benchmarks)
bytes_sent["files"] = bytes_sent["benchmark"]
bytes_sent["benchmark"] =  "multileader non-optimistic-2-buckets"
bytes_sent["buckets"] = 2 

bytes_sent["num_peers"] = bytes_sent["num_peers"]/bytes_sent["number_of_clusters"]
bytes_sent_in_gbps = (bytes_sent.groupby(["num_peers","vallen","role","number_of_clusters","cluster_batch_size","benchmark","files"])[["bytes_sent"]].sum() * 8 * 10**(-9) * 15**(-1)).reset_index()
bytes_sent_in_gbps["bytes_sent"] = bytes_sent_in_gbps["bytes_sent"] / bytes_sent_in_gbps["number_of_clusters"]
bytes_sent_in_gbps["files"] = bytes_sent_in_gbps["benchmark"]
bytes_sent_in_gbps["benchmark"] =  "multileader non-optimistic-2-buckets"
bytes_sent_in_gbps["buckets"] = 2 
```

```python
folder = "../../aws/benchmark/out/geobft/final-multileader-geobft-with-non-optimistic-global-sharing-4-buckets/" 
benchmarks = glob(f"{folder}*/")
```

```python

throughputs4 = process_throughput_benchmarks(benchmarks)
throughputs4["files"] = throughputs4["benchmark"]
throughputs4["benchmark"] = "multileader non-optimistic-4-buckets"
throughputs4["buckets"] = 4 
throughputs4["num_peers"] = throughputs4["num_peers"]/throughputs4["number_of_clusters"]
processed_cpu_usage4 = process_cpu_usage(benchmarks)
processed_cpu_usage4["num_peers"] = processed_cpu_usage4["num_peers"]/processed_cpu_usage4["number_of_clusters"]
processed_cpu_usage4["files"] = processed_cpu_usage4["benchmark"]
processed_cpu_usage4["benchmark"] = "multileader non-optimistic-4-buckets"
processed_cpu_usage4["buckets"] = 4 

bytes_sent4 = process_bytes_sent(benchmarks)
bytes_sent4["files"] = bytes_sent4["benchmark"]
bytes_sent4["benchmark"] =  "multileader non-optimistic-4-buckets"
bytes_sent4["buckets"] = 4 

bytes_sent4["num_peers"] = bytes_sent4["num_peers"]/bytes_sent4["number_of_clusters"]
bytes_sent_in_gbps4 = (bytes_sent4.groupby(["num_peers","vallen","role","number_of_clusters","cluster_batch_size","benchmark","files"])[["bytes_sent"]].sum() * 8 * 10**(-9) * 15**(-1)).reset_index()
bytes_sent_in_gbps4["bytes_sent"] = bytes_sent_in_gbps4["bytes_sent"] / bytes_sent_in_gbps4["number_of_clusters"]
bytes_sent_in_gbps4["files"] = bytes_sent_in_gbps4["benchmark"]
bytes_sent_in_gbps4["benchmark"] =  "multileader non-optimistic-4-buckets"
bytes_sent_in_gbps4["buckets"] = 4 
```

```python
throughputs = pd.concat([throughputs,throughputs4],ignore_index=True)
processed_cpu_usage = pd.concat([processed_cpu_usage,processed_cpu_usage4],ignore_index=True)
bytes_sent_in_gbps = pd.concat([bytes_sent_in_gbps,bytes_sent_in_gbps4],ignore_index=True)
bytes_sent = pd.concat([bytes_sent,bytes_sent4],ignore_index=True)
```

```python
previous_throughputs = pd.read_json(f"outputs/03-geobft-alternating/throughputs.json",orient="records")
notaltnonopt = previous_throughputs.query("benchmark == 'not-alternating-non-optimistic' and alternating == False")
previous_throughputs = previous_throughputs.groupby(["number_of_clusters","cluster_batch_size","num_peers","vallen"]).max().reset_index().query("cluster_batch_size < 256")
previous_throughputs["benchmark"] = "max"
previous_throughputs = pd.concat([previous_throughputs,notaltnonopt],ignore_index=True)
previous_throughputs["buckets"] = 1 
```

```python
throughputs = pd.concat([throughputs,previous_throughputs],ignore_index=True)
```

```python
grouped = throughputs.query("cluster_batch_size < 256").groupby(["num_peers","number_of_clusters"])
for gr, g in grouped:
    fig  = go.Figure()
    for v,data in g.sort_values(["cluster_batch_size","vallen","benchmark"]).groupby(["vallen","buckets","benchmark"]):
        # different dashes for benchmarks 
        # display(data)
        
        fig.add_trace(go.Scatter(
            name=f"vallen: {v[0]} - buckets: {v[1]} - {v[2]}",
            x=data["cluster_batch_size"],
            y=data["throughput"],
            mode='lines+markers',
            line=dict(dash='solid' if v[1]==2 else 'dash'),
            marker=dict(symbol='circle' if v[1]==2 else 'square'),
            
            
        ))
    fig.update_layout(
        title=f"GeoBFT Throughput vs Cluster Batch Size - Num Peers per Cluster: {gr[0]} - Number of Clusters: {gr[1]}",
        xaxis_title="Cluster Batch Size",
        yaxis_title="Throughput (tx/s)",
        barmode='group'
    )
    fig.show()
    tikzplotly.save(filepath=f"outputs/{notebook_name}/geobft-throughput-numpeers-{gr[0]}-clusterbatchsize-{gr[1]}.tex",fig=fig)    
```

```python
cpuusage = pd.read_json(f"outputs/03-geobft-alternating/cpu_usage.json",orient="records")
cpuusage = cpuusage.query("benchmark == 'not-alternating-non-optimistic' and alternating == False")
# cpuusage = pd.concat([cpuusage,notaltnonopt],ignore_index=True)
cpuusage["buckets"] = 1 

grouped = pd.concat([processed_cpu_usage,cpuusage]).query("cluster_batch_size < 256 and role > 0 ").groupby(["num_peers","number_of_clusters","role"])
for gr, g in grouped:
    fig  = go.Figure()
    for v,data in g.sort_values(["cluster_batch_size","vallen","benchmark"]).groupby(["vallen","benchmark","buckets"]):
        # different dashes for benchmarks 
        # display(data)
        
        fig.add_trace(go.Scatter(
            name=f"vallen: {v[0]} - {v[1]} - buckets: {v[2]}",
            x=data.groupby(["benchmark","cluster_batch_size"])["cluster_batch_size"].mean(),
            y=data.groupby(["benchmark","cluster_batch_size"])["cpu_usage"].mean(),
            error_y=dict(
                type='data',
                array=data.groupby(["benchmark","cluster_batch_size"])["cpu_usage"].std(),
                visible=True
            ),
            mode='lines+markers',
            line=dict(dash='solid' if v[2]==2 else ('dash' if v[2]==4 else 'dot'))
        ))
    fig.update_layout(
        title=f"GeoBFT - cpu_usage - numpeers={gr[0]} - numberofclusters={gr[1]} - role={gr[2]}",
        xaxis_title="Cluster Batch Size",
        yaxis_title="cpu_usage in percent",
        barmode='group'
    )
    fig.show()
    tikzplotly.save(filepath=f"outputs/{notebook_name}/geobft-cpuusage-numpeers-{gr[0]}-clusterbatchsize-{gr[1]}-role-{gr[2]}.tex",fig=fig)
```
